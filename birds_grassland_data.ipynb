```python
# Bird monitoring grassland data.

import pandas as pd

file_path = r"c:\Users\Ravi\Downloads\Bird_Monitoring_Data_GRASSLAND.XLSX"     # Given data.

excel_file = pd.ExcelFile(file_path)                      # Open the data using pandas.

all_sheets = pd.read_excel(file_path, sheet_name=None)    # Read the excel file using pandas.

print(excel_file.sheet_names)
```

```python
# Combined all sheets and make a dataframe using pandas.

combined_df = pd.concat(all_sheets.values(), ignore_index=True)   

df = combined_df
```

```python
# Checking the duplicated rows.

df.duplicated().sum()
```

```python
# Find the scientific names of the null values.

df_id = df[df['ID_Method'].isna()]['Scientific_Name']

df_acc = df[df['AcceptedTSN'].isna()]['Scientific_Name']

df_tax = df[df['TaxonCode'].isna()]['Scientific_Name']


print(df_id)
print(df_acc)
print(df_tax)
```

```python
# Find the null using scientific names.

df[df['Scientific_Name'] == 'Haemorhous mexicanus']['AcceptedTSN'].unique()
```

```python
# Filling the null values with alternatives.

df['Sub_Unit_Code'] = df['Sub_Unit_Code'].fillna('Not having')   # By assumption.  
df['ID_Method'] = df['ID_Method'].fillna('Visualization')        # From scientific name.
df['Distance'] = df['Distance'].fillna('>=100 Meters')           # By assumption.
df['AcceptedTSN'] = df['AcceptedTSN'].fillna('815358')           # From scientific name.
df.loc[df['Scientific_Name'] =='Corvus caurinus','TaxonCode'] = df.loc[df['Scientific_Name'] == 
                                                                'Corvus caurinus','TaxonCode'].fillna(35821)
df.loc[df['Scientific_Name'] == 'Ardeola bacchus','TaxonCode'] = df.loc[df['Scientific_Name'] ==
                                                                'Ardeola bacchus','TaxonCode'].fillna(18435)
                                                                 # From scientific name.
df = df.drop_duplicates().reset_index(drop=True)
df.isna().sum()
```

```python
# Insert the cleaned data into the database.

from sqlalchemy import create_engine

engine = create_engine("postgresql://postgres:12345678@localhost:5432/birds_monitoring")

df.to_sql('bird_grassland_data', engine, if_exists='replace', index=False)

print("Data inserted successfully into the database.")
```
