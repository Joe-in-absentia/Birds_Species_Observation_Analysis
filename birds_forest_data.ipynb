```python
# Birds monitoring forest data.

import pandas as pd

file_path = r"C:\Users\Ravi\Downloads\Bird_Monitoring_Data_FOREST.xlsx"   # Given data.

excel_file = pd.ExcelFile(file_path)                         # Open excel file using pandas.

all_sheets = pd.read_excel(file_path, sheet_name=None)       # Read the excel file using pandas.

print(excel_file.sheet_names)
```

```python
# Combined the all sheets and make dataframe using pandas.

combined_df = pd.concat(all_sheets.values(), ignore_index=True)

df = combined_df
```

```python
# Checking the null values.

df.isnull().sum()
```

```python
# Checking the duplicated rows.

df.duplicated().sum()
```

```python
# Find the scientific name of the null values.

df_id = df[df['ID_Method'].isna()]['Scientific_Name']

df_acc = df[df['AcceptedTSN'].isna()]['Scientific_Name']

print(df_id)
print(df_acc)
```

```python
# Find the null values using scientific name.

df[df['Scientific_Name'] == 'Haemorhous mexicanus']['AcceptedTSN'].unique()
```

```python
# Filling the null values with alternatives.

df['Sub_Unit_Code'] = df['Sub_Unit_Code'].fillna('Not having')   # By assumption.  
df['ID_Method'] = df['ID_Method'].fillna('Singing')              # From scientific name.
df['Distance'] = df['Distance'].fillna('>=100 Meters')           # By assumption.
df['Sex'] = df['Sex'].fillna('Undetermined')                     # By assumption.
df['AcceptedTSN'] = df['AcceptedTSN'].fillna('815358')           # From scientific name.

df = df.drop_duplicates().reset_index(drop=True)
df.isna().sum()
```

```python
# Insert the cleaned data into the database.

from sqlalchemy import create_engine

engine = create_engine("postgresql://postgres:12345678@localhost:5432/birds_monitoring")

df.to_sql('bird_forest_data', engine, if_exists='replace', index=False)

print("Data inserted successfully into the database.")
```
